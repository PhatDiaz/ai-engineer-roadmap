**Ex 2.** Cho:

numbers = [4, 2, 9, 1, 7]

1.   Tìm số lớn nhất
2.   Tìm số nhỏ nhất
3.   Tính tổng các phần tử.
4.   Tính trung bình
5.   Sắp xếp tăng dần
6.   Sắp xếp giảm dần

```python
numbers = [4, 2, 9, 1, 7]
temp_min = numbers[0]
temp_max = numbers[1]
sum = 0

for number in numbers:
  if number >= temp_max:
    temp_max = number
  elif number <= temp_min:
    temp_min = number

# 1. Tìm số lớn nhất
print(temp_max)

# 2. Tìm số nhỏ nhất
print(temp_min)

for number in numbers:
  sum += number

# 3. Tính tổng các phần tử
print(sum)

# 4. Tính trung bình
average = sum / len(numbers)
print(average)

# 5. Sắp xếp tăng dần
print(sorted(numbers))

# 6. Sắp xếp giảm dần
print(sorted(numbers, reverse=True))
```

# **Task 5 — Mini challenge**

**Hãy tự viết chương trình xử lý dữ liệu:**

students = [

    {"name": "An", "score": 8.5},
    {"name": "Binh", "score": 6.0},
    {"name": "Chi", "score": 9.0},
    {"name": "Dung", "score": 7.5},
    {"name": "Hoa", "score": 5.5}
]

**Chương trình phải tìm được:**

1. Điểm trung bình của lớp
2. Sinh viên có điểm cao nhất
3. Sinh viên có điểm thấp nhất
4. Danh sách sinh viên có score >= 8
5. Sắp xếp sinh viên theo score giảm dần

**Output mong đợi tương tự:**

Average score: 7.3

Highest:
Chi - 9.0

Lowest:
Hoa - 5.5

Students >= 8:
An - 8.5;
Chi - 9.0

Ranking
Chi - 9.0;
An - 8.5;
Dung - 7.5;
Binh - 6.0;
Hoa - 5.5

```python
students = [
    {"name": "An", "score": 8.5},
    {"name": "Binh", "score": 6.0},
    {"name": "Chi", "score": 9.0},
    {"name": "Dung", "score": 7.5},
    {"name": "Hoa", "score": 5.5}
]

total_score = 0
for student in students:
  total_score += student["score"]

# 1. Điểm trung bình của lớp
print("Average score:", total_score / len(students))
print("\n")

max_score = students[0]["score"]
min_score = students[0]["score"]

for student in students:
  if student["score"] >= max_score:
    max_score = student["score"]
  elif student["score"] <= min_score:
    min_score = student["score"]

for student in students:
  if student["score"] == max_score:

# 2. Sinh viên có điểm cao nhất
    print("\n")
    print("Highest: ")
    print(student["name"], "-", student["score"])

  elif student["score"] == min_score:

# 3. Sinh viên có điểm thấp nhất
    print("\n")
    print("Lowest: ")
    print(student["name"], "-", student["score"])

# 4. Danh sách sinh viên có score >= 8
print("\n")
print("Students >= 8: ")
for student in students:
  if student["score"] >= 8:
     print(student["name"], "-", student["score"])

# 5. Sắp xếp sinh viên theo score giảm dần
print("\n")
print("Ranking: ")

ranking = sorted(
    students,
    key = lambda student: student["score"],
    reverse = True
)

for student in ranking:
  print(student["name"], "-", student["score"])
```
