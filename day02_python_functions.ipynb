# **Task 5 — List Comprehension**

```python
# Cú pháp tổng quát: [expression for item in iterable]

numbers = [1, 2, 3, 4, 5]

squares = []

# Thay vì:
# for number in numbers:
#   squares.append(number ** 2)

# Ta viết:
squares = [number ** 2 for number in numbers]
print(squares)
```

# **Task 7 — Mini Challenge**

```python
students = [
    {"name": "An", "score": 8.5},
    {"name": "Binh", "score": 6.0},
    {"name": "Chi", "score": 9.0},
    {"name": "Dung", "score": 7.5},
    {"name": "Hoa", "score": 5.5}
]

# trả về điểm trung bình
def calculate_average(students):
  sum = 0
  for student in students:
    sum += student["score"]
  return sum / len(students)

print(calculate_average(students))

# trả về dictionary của sinh viên điểm cao nhất
def get_highest_student(students):
  return max(
      students,
      key = lambda student: student["score"]
  )

print(get_highest_student(students))

# trả về những sinh viên: score >= 6.5
def get_passed_students(students):
  return [student for student in students if student["score"] >= 6.5]
print(get_passed_students(students))


def get_students_above_score(students, min_score):
  return [student["name"] for student in students if student["score"] >= min_score]
print(get_students_above_score(students, 8))


def rank_students(students):
  return sorted(
      students,
      key = lambda student: student["score"],
      reverse = True
  )
print(rank_students(students))


def get_passed_count(students):
  return len([student for student in students if student["score"] >= 6.5])

def get_highest_score(students):
  return max(student["score"] for student in students)

def get_lowest_score(students):
  return min(student["score"] for student in students)

def get_statistics(students):
  return {
      "average": calculate_average(students),
      "highest": get_highest_score(students),
      "lowest": get_lowest_score(students),
      "passed_count": get_passed_count(students)
  }

print(get_statistics(students))
```
