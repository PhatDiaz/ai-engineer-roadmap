# **Task 1 — Pandas là gì?**

````python
# Import
import pandas as pd
import numpy as np
````

**NumPy Array**

```
scores = np.array([
    [8.5, 7.0, 9.0],
    [6.0, 8.0, 7.5]

```

Dữ liệu:
```
[[8.5 7.0 9.0]

Bạn phải tự nhớ:
```

column 2 = AI

**Pandas DataFrame**

````python
data = {
    "Math": [8.5, 6.0],
    "English": [7.0, 8.0],
    "AI": [9.0, 7.5]
}

df = pd.DataFrame(data)

print(df)
````

Pandas có column names nên dễ làm việc với dataset thực tế hơn

# **Task 2 — Series**

Series có thể hiểu là một cột dữ liệu 1 chiều

````python
# Ví dụ
scores = pd.Series([8.5, 7.0, 9.0, 6.0])

print(scores)
````

**Series có index riêng**

````python
scores = pd.Series(
    [8.5, 7.0, 9.0],
    index=["An", "Binh", 'Cuong']
)

print(scores)

print(scores['An'])
````

**Một số thuộc tính Series**

```
print(scores.shape)
print(scores.size)
print(scores.dtype)
print(scores.index)
print(scores.values)
```

**Series vs NumPy Array**

````python
# NumPy
arr = np.array([8.5, 7.0, 9.0])

# Pandas
series = pd.Series([8.5, 7.0, 9.0])
````

**Điểm khác nổi bật:**

**NumPy Array**
→ chủ yếu dựa trên position

**Pandas Series**
→ có index/label

# **Task 3 — DataFrame**

**DataFrame = bảng gồm nhiều rows và columns**

````python
# Ví dụ
students = {
    "Name": ["An", "Binh", "Cuong", "Dung"],
    "Age": [22, 25, 21, 23],
    "Score": [8.5, 7.0, 9.0, 6.0]
}

df = pd.DataFrame(students)
print(df)
````

**Hiểu cấu trúc DataFrame**

```
       columns
          ↓
     Name Age Score
0     An   22  8.5
1   Binh   25  7.0
2  Cuong   21  9.0
↑
index
```
DataFrame có:

```
rows
columns
index
values
data types
```

**df.shape**

```
print(df.shape)
```
→ (4, 3)

Nghĩa là:

```
4 rows
3 columns
```

**df.columns**

```
print(df.columns)
```
→ đại loại:

```
Index(['Name', 'Age', 'Score'], dtype='object')
```

**df.index**

```
print(df.index)
```
→

```
RangeIndex(start=0, stop=4, step=1)
```

**df.dtypes**

```
print(df.dtypes)
```
Có thể ra:

```
Name      object
Age        int64
Score    float64
```

Điểm rất quan trọng:

```
NumPy array
→ thường có một dtype chung

Pandas DataFrame
→ mỗi column có thể có dtype riêng
```

Ví dụ:
```
Name  → string/object
Age   → int
Score → float
```

**Lấy một column**

```
print(df["Name"])
```

Kết quả là Series:

```
0       An
1     Binh
2    Cuong
3     Dung
```

**Lấy nhiều columns**

```
print(
    df[
        ["Name", "Score"]
    ]

```
Kết quả là DataFrame:

```

```
Đây là điểm dễ nhầm:

`df["Score"]`

**→ Series**

# **Task 4 — Đọc CSV**

# **Task 5 — Khám phá Dataset**

Đây là phần cực kỳ quan trọng trong AI/ML.

Khi nhận một dataset mới, **đừng bắt đầu train model ngay**.

Việc đầu tiên thường là hiểu dữ liệu.

Một workflow đơn giản:

```
Load Dataset
     ↓
head()
     ↓
shape
     ↓
columns
     ↓
dtypes
     ↓
info()
     ↓
describe()
```

**head()**

`df.head()`

Mặc định:

`5 rows đầu`

Nếu:

`df.head(3)`

→ 3 rows đầu

**tail()**

Mặc dù không phải trọng tâm roadmap hôm nay, bạn nên biết:

`df.tail()`

→ 5 rows cuối

**shape**

`df.shape`

Ví dụ:
`(5, 4)`

→
5 rows, 4 columns

**columns**

`df.columns`

→ tên các columns.

**dtypes**

`df.dtypes`

→ dtype của từng column

**info()**

`df.info()`

Đây là function rất quan trọng.

Nó cho ta biết:

```
số rows
columns
non-null count
dtype
memory
```
Ví dụ:

```
Name        5 non-null        object
Age         5 non-null        int64
Score       5 non-null        float64
City        5 non-null        object
```

Sau này khi dataset có missing values:


```
1000 rows

Age
950 non-null
```
bạn sẽ phát hiện:

`50 missing values`

**describe()**

`df.describe()`

Thường áp dụng cho numerical columns

**Ví dụ:**
```
            Age      Score

count       5.0       5.0
mean        ...
std         ...
min         ...
25%         ...
50%         ...
75%         ...
max         ...
```
Ta đã học: **mean, std, min, max** ở NumPy

Bây giờ Pandas tổng hợp chúng cho từng numerical column

**25%, 50%, 75% là gì?**

Chúng là **quartiles** (tứ phân vị)

Chỉ cần biết là:

```
25% → Q1
50% → median
75% → Q3
```

# **Task 6 — Coding Practice**

````python
# Ex 2: Custom index

scores = pd.Series([8.5, 7.0, 9.0, 6.0], index=["An", "Binh", "Cuong", "Dung"])

print(scores["Cuong"])
````

````python
# Ex 3: DataFrame

data = {
    "Name": ["An", "Binh", "Cuong", "Dung"],
    "Age": [22, 25, 21, 23],
    "Score": [8.5, 7.0, 9.0, 6.0]
}

scores = pd.DataFrame(data)

print(scores)

# Ex 4
print(scores.shape)

print(scores.columns)

print(scores.index)

print(scores.dtypes)

# Ex 5
print(scores["Score"])
print(type(scores["Score"]))

# Ex 6
print(scores[["Name", "Score"]])
print(type(scores[["Name", "Score"]]))

# Ex 7
scores.to_csv("students.csv", index=False)

# Ex 8
print(pd.read_csv("students.csv"))

# Ex 9
print(scores.head(3))

# Ex 10
print(scores.tail(2))

# Ex 11
print(scores.info())
# Bài 11 print ra
# - Có 4 rows
# - Có 3 columns
# - column age là int
# - column score là float

# Ex 12
print(scores.describe())
# mean của Age là 22.750000
# mean của Score là 7.625000
# min Score là 6.000000
# max Score là 9.000000

# Ex 13
print(scores["Score"].mean())

# Ex 14
print(scores["Score"].max())
````

````python
# Ex 15

scores_numpy = np.array([8.5, 7.0, 9.0, 6.0])

scores_pandas = pd.Series([8.5, 7.0, 9.0, 6.0])

print(np.mean(scores_numpy))
print(scores_pandas.mean())

# Kết quả mean giống nhau vì cùng dữ liệu. NumPy array chủ yếu làm việc dựa trên vị trí và mạnh về numerical computation,
# còn Pandas Series có index/label nên thuận tiện hơn khi xử lý dữ liệu dạng bảng.
````

# **Task 7 — Mini Challenge**

````python
import pandas as pd

data = {
    "Name": ["An", "Binh", "Cuong", "Dung", "Hoa", "Huy"],
    "Age": [22, 25, 21, 23, 24, 26],
    "Math": [8.5, 6.0, 9.0, 5.5, 7.5, 8.0],
    "English": [7.0, 8.0, 9.5, 6.0, 8.5, 7.5],
    "AI": [9.0, 7.5, 8.5, 7.0, 8.0, 9.5],
    "City": ["Hanoi", "HCM", "Hanoi", "Danang", "Hanoi", "HCM"]
}

df = pd.DataFrame(data)

# Require 1: Overview
display(df)

print(df.shape)

print(df.columns)

print(df.dtypes)

# Require 2: First Rows
display(df.head(3))

# Require 3: Dataset Information
print(df.info())

# Có 6 rows, 6 columns và không có missing value

# Require 4: Statistics
print(df.describe())

# Average Age: 23.500000
# Average Math: 7.416667
# Average English: 7.750000
# Average AI: 8.250000

# Require 5: Select Column
print(df["Name"])
# -> Kết quả là Series vì là cột dữ liệu 1 chiều

# Require 6: Select Multiple Columns
print(df[["Name", "Math", "AI"]])
# -> Kết quả là DataFrame

# Require 7: Subject Average
print(df[["Math", "English", "AI"]].mean())

# Require 8: Highest / Lowest
print(df[["Math", "AI"]].max())
print(df[["Math", "AI"]].min())

# Require 9: Save CSV
df.to_csv("student_dataset.csv", index=False)

# Require 10: Load CSV
loaded_df = pd.read_csv("student_dataset.csv")
loaded_df.head()

# Optional Challenge
df["Average"] = (df["Math"] + df["English"] + df["AI"]) / 3
df.head()
````
