# **Task 2 — Element-wise Operations**

````python
# Cho
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

# Cộng từng phần tử tương ứng của 2 array
print(a + b)

# Trừ từng phần tử tương ứng của 2 array
print(a - b)

# Nhân từng phần tử tương ứng của 2 array
print(a * b)

# Chia từng phần tử tương ứng của 2 array
print(a / b)

# Bình phương từng phần tử của array A
print(a ** 2)
````

# **Task 3 — Dot Product**
# **(Tích vô hướng)**

````python
# Cho
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

# Tính: 1×4 + 2×5 + 3×6
print(np.dot(a, b))

# Có thể viết a @ b
print(a @ b)
````

**Matrix multiplication**

````python
# Cho
A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])

print(A @ B)
````

**Vì sao dot product quan trọng trong AI?**



Sau này Neural Network có dạng:

`y = XW + b`

Trong code:

`output = X @ W + b`

Nên:

`matrix multiplication + broadcasting`

sẽ xuất hiện liên tục.

# **Task 4 — Hiểu axis**

**Cho:**

```
scores = np.array([    
  [8, 7, 9],    
  [6, 8, 7],    
  [9, 9, 8]
])
```
Có thể hiểu:
```
          Math   English   AI
Student1    8       7      9
Student2    6       8      7
Student3    9       9      8
```

**Không dùng axis**

`np.mean(scores, axis=1)`

NumPy tính trung bình toàn bộ 9 giá trị

**axis = 0**

`np.mean(scores, axis=0)`

```
Output:
[7.666... 8.0 8.0]
```
**Hiểu là: đi theo chiều dọc, gộp các row lại**

Kết quả là trung bình từng column.

**Tức là**

```
Math average
English average
AI average
```

**axis = 1**

`np.mean(scores, axis=1)`
```
Output đại loại:
[8.0 7.0 8.666...]
```
**Hiểu là: tính theo từng row**

```
Tức là:
Student 1 average
Student 2 average
Student 3 average
```

**Mẹo nhớ**

**Với matrix:**
```
axis=0
→ collapse rows
→ còn columns
```
```
axis=1
→ collapse columns
→ còn rows
```

**Hoặc trong ví dụ student:**
```
axis=0 → theo môn

axis=1 → theo student
```

# **Task 5 — Statistics**

````python
scores = ([8, 7, 9, 6])

# Sum
print(np.sum(scores))

# Mean
print(np.mean(scores))

# Min
print(np.min(scores))

# Max
print(np.max(scores))

# Standard Deviation: độ phân tán của dữ liệu quanh mean
print(np.std(scores))
````

# **Task 7 — Mini Challenge**

# **Student Score Analysis with NumPy**

````python
scores = np.array([
    [8.5, 7.0, 9.0],
    [6.0, 8.0, 7.5],
    [9.0, 9.5, 8.5],
    [5.5, 6.0, 7.0]
])

# Require 1: Average toàn bộ
overall_average = np.mean(scores)
print(overall_average)

# Require 2: Average từng Student
print(np.mean(scores, axis = 1))

# Require 3: Average từng môn
print(np.mean(scores, axis = 0))

# Require 4: Highest score
print(np.max(scores))

# Require 5: Lowest score
print(np.min(scores))

# Require 6: Highest score từng student
print(np.max(scores, axis = 1))

# Require 7: Standard deviation từng môn
print(np.std(scores, axis = 0)) # môn nào có std lớn hơn nghĩa là điểm giữa các student phân tán hơn

# Require 8: Best student
average_students = np.mean(scores, axis = 1)
print(np.argmax(average_students) + 1)

# Require 9: Best subject
average_subjects = np.mean(scores, axis = 0)
print(np.argmax(average_subjects) + 1)

# Require 10: Weighted Score

weights = np.array([0.3, 0.2, 0.5])
weighted_scores = scores @ weights
print(weighted_scores)

# Optional Challenge
print(np.argmax(weighted_scores) + 1)
````
