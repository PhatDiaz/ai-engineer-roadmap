# **Task 1 — Class & Object**

Trước đây, chúng ta quản lý student bằng dictionary:

````python
student = {
    "name": "An",
    "score": 9.0
}
````

OOP cho phép tạo 1 kiểu dữ liệu riêng:

```
class Student:
  pass
```



Sau đó:

```
student1 = Student()
student2 = Student()
```



Ở đây:



```
Student   → Class

student1  → Object
student2  → Object
```



Hãy hình dung:

```
Class = bản thiết kế

Object = sản phẩm được tạo từ bản thiết kế
```



**Attribute**

````python
# Ta có thể gắn dữ liệu vào object

class Student:
  pass

student1 = Student()

student1.name = "An"
student1.score = 9.0

print(student1.name)
print(student1.score)

# name & score gọi là attributes
````

# **Task 2 — __init__ và self**

Thay vì:


```
student1 = Student()

student1.name = "An"
student1.score = 8.5
```



Ta dùng constructor:


```
class Student:
  def __init__(self, name, score):
    self.name = name
    self.score = score
```

Tạo object:


```
student1 = Student("An", 8.5)
student2 = Student("Binh", 6.0)
```



**__init__ là gì?**

__init__ được gọi tự động khi tạo object.

Khi bạn viết:

```student1 = Student("An", 8.5)```

Python sẽ gọi:

```__init__(self, "An", 8.5)```

Và lưu:

```
self.name = "An"
self.score = 8.5
```



**self là gì?**

Trong:


```
class Student:

    def __init__(self, name, score):
        self.name = name
        self.score = score
```

--> Self đại diện cho object hiện tại

Ví dụ:


```
student1 = Student("An", 8.5)
```

thì về mặt tư duy:

`self → student1`

nên:

`self.name = name`

có thể hiểu như:

`student1.name = "An"`

Sau đó:

`student2 = Student("Binh", 6.0)`

lúc này:

`self → student2`

# **Task 3 — Methods**

**Method là function nằm bên trong class.**

````python
# Ví dụ

class Student:
  def __init__(self, name, score):
    self.name = name
    self.score = score

  def display_info(self):
    print(self.name, "-", self.score)

# Gọi
student1 = Student("An", 8.5)
student1.display_info()
````

**Method có return**

````python
# Ví dụ

class Student:
  def __init__(self, name, score):
    self.name = name
    self.score = score

  def is_passed(self):
    return self.score >= 6.5

# Gọi
student1 = Student("An", 8.5)
print(student1.is_passed())
````

**Attribute vs Method**

Nhìn:

`student.name`

đây là:

`attribute`

Nhưng:

`student.is_passed()`

đây là:

`method`

Dấu () rất quan trọng.

**Dictionary vs Object**

*Nhớ:*

```
Dictionary
→ student["score"]

Object
→ student.score
```



# **Task 4 — Inheritance**

Inheritance = class con kế thừa dữ liệu/hành vi từ class cha

````python
# Ví dụ

class Person:
  def __init__(self, name, age):
    self.name = name
    self.age = age

# Tạo class Student
class Student(Person):
  def __init__(self, name, age, score):
    super().__init__(name, age)

    self.score = score

# Student có name, age, score. Trong đó: name, age được kế thừa từ Person
````

**super() là gì?**

Dòng:

`super().__init__(name, age)`

nghĩa là:

Gọi constructor __init__() của class cha

Nếu không dùng inheritance, chúng ta có thể viết lại:

```
self.name = name
self.age = age
```

Nhưng:

`super().__init__(name, age)`

giúp chúng ta tái sử dụng code


# **Task 5 — Exception Handling**

Chương trình:

`number = int(input("Enter number: "))`

Nếu user nhập:

`abc`

Python sẽ báo:

`ValueError`

và chương trình dừng.

````python
# Ta có thể xử lý
try:
  number = int(input("Enter number: "))

except ValueError:
  print("Invalid number")
````

**Cấu trúc try / except**

```
try:
    # đoạn code có thể gây lỗi

except:
    # xử lý lỗi
```

Nhưng không nên dùng *except: chung chung* nếu biết loại lỗi.

**Một số exception cần biết hôm nay**

`int("abc")`   →    ValueError

`10 / 0`       →    ZeroDivisionError


```
numbers = [1, 2, 3]

print(numbers[10])

→ IndexError
```


```
student = {
    "name": "An"
}

print(student["score"])

→ KeyError
```



**Else và Finally**

````python
# Ví dụ
try:
    number = int(input("Enter number: "))

except ValueError:
    print("Invalid input")

else:
    print("Valid number:", number)

finally:
    print("Program finished")
````

**Luồng:**

```
try
 ↓
Error?
 ├── Yes → except
 │
 └── No → else

finally
↓
luôn chạy
```

````python
# Ex 2

class Rectangle:
  def __init__(self, width, height):
    self.width = width
    self.height = height

  def area(self):
    return self.width * self.height

  def perimeter(self):
    return 2 * (self.width + self.height)

rectangle1 = Rectangle(4, 5)
print(rectangle1.area())
print(rectangle1.perimeter())
````

# **Task 7 — Mini Challenge**

````python
class Student:
  def __init__(self, name, score):
    self.name = name
    self.set_score(score)

  def set_score(self, score):
    if score < 0 or score > 10:
      raise ValueError("Score must be between 0 and 10")
    self.score = score

  def is_passed(self):
    return self.score >= 6.5

  def get_grade(self):
    if self.score >= 8.5 :
      return "Excellent"
    elif self.score >= 7 :
      return "Good"
    elif self.score >= 5 :
      return "Average"
    else:
      return "Fail"

  def display_info(self):
    print(f"{self.name} - {self.score} - {self.get_grade()}")


class StudentManager:
  def __init__(self):
    self.students = []

  def add_student(self, student):
    self.students.append(student)

  def display_students(self):
    for student in self.students:
      student.display_info()

  def update_student_score(self, name, new_score):
    for student in self.students:
      if student.name == name:
        student.set_score(new_score)
        return
    print("Student not found")

  def delete_student(self, name):
    for student in self.students:
      if student.name == name:
        self.students.remove(student)
        return
    print("Student not found")

  def get_highest_student(self):
    if not self.students:
      return None

    return max(
        self.students,
        key = lambda student: student.score
    )

  def calculate_average(self):
    if not self.students:
      return None

    return sum(student.score for student in self.students) / len(self.students)

manager = StudentManager()
manager.add_student(Student("An", 8.5))
manager.add_student(Student("Binh", 7.0))
manager.add_student(Student("Cuong", 5.0))
manager.display_students()

print("After delete: ")
manager.delete_student("Cuong")
manager.display_students()
manager.calculate_average()
````
