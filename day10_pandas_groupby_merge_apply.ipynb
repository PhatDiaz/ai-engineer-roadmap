# **Task 1 — value_counts()**

````python
# Dataset
import pandas as pd

data = {
    "Name": ["An", "Binh", "Cuong", "Dung", "Hoa", "Huy"],

    "City": ["Hanoi", "HCM", "Hanoi", "Danang", "Hanoi", "HCM"],

    "Score": [8.5, 7.0, 9.0, 6.0, 8.0, 7.5]
}

df = pd.DataFrame(data)

# Muốn biết mỗi thành phố có bao nhiêu student:
print(df["City"].value_counts())

#  value_counts()
#       ↓
# đếm frequency của từng giá trị unique
````

**value_counts(normalize=True)**

````python
# Optional nhưng nên biết:
print(df["City"].value_counts(normalize=True))

# -> Tỷ lệ thay vì số lượng
````

**unique() và nunique()**

````python
# Để hỗ trợ hiểu value_counts():
print(df["City"].unique())
# -> Các giá trị khác nhau

# Còn:
print(df["City"].nunique())
# -> Số lượng giá trị unique
````

***NHỚ***
```
unique()
→ các giá trị unique là gì?

nunique()
→ có bao nhiêu giá trị unique?

value_counts()
→ mỗi giá trị xuất hiện bao nhiêu lần?
```

# **Task 2 — groupby()**

````python
# Dataset
data = {
    "Name": ["An", "Binh", "Cuong", "Dung", "Hoa", "Huy"],

    "City": ["Hanoi", "HCM", "Hanoi", "Danang", "Hanoi", "HCM" ],

    "Score": [8.5, 7.0, 9.0, 6.0, 8.0, 7.5]
}

df = pd.DataFrame(data)

# Muốn tính average Score theo City:
print(df.groupby("City")["Score"].mean())
````

**Hiểu groupby()**

Tư duy:
```
Dataset
   ↓
chia thành các group theo City

Hanoi
→ An
→ Cuong
→ Hoa

HCM
→ Binh
→ Huy

Danang
→ Dung
```

Sau đó:
```
mỗi group
   ↓
mean()
```

Tóm lại:
```
Split
↓
Apply
↓
Combine
```

**Group theo một column**

`df.groupby("City")["Score"].mean()`

Phân tích:
```
df.groupby("City")
→ nhóm rows theo City

["Score"]
→ chỉ lấy Score

.mean()
→ tính mean từng group
```



````python
# Các aggregation phổ biến

print(df.groupby("City")["Score"].mean())
print(df.groupby("City")["Score"].sum())
print(df.groupby("City")["Score"].min())
print(df.groupby("City")["Score"].max())
print(df.groupby("City")["Score"].count())
````

# **Task 4 — apply()**

apply() cho phép áp dụng một function lên Series hoặc DataFrame

````python
# Ví dụ:
scores = pd.Series([8.5 ,7.0, 9.0, 6.0])

def classify_score(score):
  if score >= 8.5:
    return "Excellent"
  elif score >= 7:
    return "Good"
  elif score >= 5:
    return "Average"
  else:
    return "Fail"

print(scores.apply(classify_score))
````

**apply() với DataFrame column**

```
df["Grade"] = df["Score"].apply(
    classify_score
)
```
Bây giờ:
```
Name    Score    Grade
An      8.5      Excellent
Binh    7.0      Good
...
```
Đây là pattern rất phổ biến:
```
existing column
      ↓
apply(function)
      ↓
new column
```

**apply() + lambda**

Nếu function đơn giản:
```
df["Passed"] = df["Score"].apply(
    lambda score: score >= 5
)
```
Output:
```
True
True
True
...
```




**Khi nào dùng apply()?**

Ví dụ cần logic:
```
def classify_score(score):
    ...
```
thì:

```
df["Score"].apply(
    classify_score
)
```
rất dễ đọc

Nhưng nếu chỉ muốn:

`Score * 2`

đừng viết:
```
df["Score"].apply(
    lambda x: x * 2
)
```

Ưu tiên:
```
df["Score"] * 2
```

# **Task 5 — merge()**

Trong thực tế dữ liệu có thể nằm ở nhiều bảng

````python
# Ví dụ bảng students

students = pd.DataFrame({
    "StudentID": [101, 102, 103],
    "Name": ["An", "Binh", "Cuong"]
})

# Bảng Scores
scores = pd.DataFrame({
    "StudentID": [101, 102, 103],
    "Score": [8.5, 7.0, 9.0]
})

# Muốn ghép:
merge_df = pd.merge(students, scores, on="StudentID")
print(merge_df)
````

**Hiểu on**

```
on="StudentID"
```
nghĩa là:

`Dùng StudentID làm key để ghép hai DataFrame`

Tư duy rất giống database join
```
students
StudentID
    ↓
match
    ↑
StudentID
scores
```

**inner merge**

````python
# Giả sử
students = pd.DataFrame({
    "StudentID": [101, 102, 103],
    "Name": ["An", "Binh", "Cuong"]
})

scores = pd.DataFrame({
    "StudentID": [101, 102, 104],
    "Score": [8.5, 7.0, 9.0]
})

# Nếu
pd.merge(students, scores, on="StudentID", how="inner")

# Chỉ giữ ID xuất hiện ở cả 2 bảng: 101, 102
````

**left merge**

````python
pd.merge(students, scores, on="StudentID", how="left")

# Chỉ giữ ID xuất hiện ở bảng students (giữ tất cả rows của bảng bên trái)
````

**Tóm tắt:**

```
inner
→ chỉ phần match

left
→ giữ tất cả bên trái

right
→ giữ tất cả bên phải

outer
→ giữ tất cả hai bên
```

# **Coding Practice — 15 bài**

````python
# Dùng dataset

students = pd.DataFrame({
    "StudentID": [101, 102, 103, 104, 105, 106],
    "Name": ["An", "Binh", "Cuong", "Dung", "Hoa", "Huy"],
    "City": ["Hanoi", "HCM", "Hanoi", "Danang", "Hanoi", "HCM"],
    "Math": [8.5, 6.0, 9.0, 5.5, 7.5, 8.0],
    "AI": [9.0, 7.5, 8.5, 7.0, 8.0, 9.5]
})

# Ex 1: value_counts()
print(students["City"].value_counts())

# Ex 2: Unique City
print(students["City"].unique())
print(students["City"].nunique())

# Ex 3: Group Mean
print(students.groupby("City")["Math"].mean())

# Ex 4: Group AI
print(students.groupby("City")["AI"].mean())

# Ex 5: Multiple Columns
print(students.groupby("City")[["Math", "AI"]].mean())

# Ex 6: Multiple Aggregations
print(students.groupby("City")["Math"].agg(["mean", "min", "max"]))

# Ex 7: Count per group
print(students.groupby("City")["Name"].count())

# Ex 8: Apply Classifications
def classify_ai(score):
  if score >= 9:
    return "Excellent"
  elif score >= 8:
    return "Good"
  elif score >= 7:
    return "Average"
  else:
    return "Fail"

students["AI_Level"] = students["AI"].apply(classify_ai)
print(students)

# Ex 9: Apply Lambda
students["Passed_AI"] = students["AI"].apply(lambda score: score >= 7)
print(students)

# Ex 10: Average Column
students["Average"] = (students["Math"] + students["AI"]) / 2
print(students)
````

**Merge Exercises**

````python
courses = pd.DataFrame({
    "StudentID": [101, 102, 103, 107],
    "Course": ["AI", "Data Science", "AI", "Machine learning"]
})

# Ex 11: Inner merge
print("------------Inner------------")
inner_merge_df = pd.merge(students, courses, on="StudentID", how="inner")
print(inner_merge_df)
print("\n")

# Ex 12: Left merge
print("------------Left------------")
print(pd.merge(students, courses, on="StudentID", how="left"))
print("\n")

# Ex 13: Outer merge
print("------------Outer------------")
print(pd.merge(students, courses, on="StudentID", how="outer"))
print("\n")

# Ex 14: Group after Merge
print(inner_merge_df.groupby("Course")["AI"].mean())
````

````python
# Ex 15: Combined Task
students["Average"] = (students["Math"] + students["AI"]) / 2

def classify_student(average):
    if average >= 8.5:
        return "Excellent"
    elif average >= 7:
        return "Good"
    elif average >= 5:
        return "Average"
    else:
        return "Fail"

students["Level"] = students["Average"].apply(classify_student)

city_average = students.groupby("City")["Average"].mean()

city_average = city_average.sort_values(ascending=False)

print(students)
print(city_average)
````

# **Task 7 — Mini Challenge**

````python
students = pd.DataFrame({
    "StudentID": [101, 102, 103, 104, 105, 106],
    "Name": ["An", "Binh", "Cuong", "Dung", "Hoa", "Huy"],
    "City": ["Hanoi", "HCM", "Hanoi", "Danang", "Hanoi", "HCM"],
})

scores = pd.DataFrame({
    "StudentID": [101, 102, 103, 104, 105, 106],
    "Math": [8.5, 6.0, 9.0, 5.5, 7.5, 8.0],
    "English": [7.0, 8.0, 9.5, 6.0, 8.5, 7.5],
    "AI": [9.0, 7.5, 8.5, 7.0, 8.0, 9.5]
})

# Require 1: Merge
df = pd.merge(students, scores, on="StudentID")
print(df)

# Require 2: Overview
# print(df.head())

# print(df.shape)

# df.info()

# Require 3: City Count
print(df["City"].value_counts())

# Require 4: Average Score
df["Average"] = (df['AI'] + df['Math'] + df['English']) / 3
print(df)

# Require 5: Classification
def classify_student(average):
  if average >= 8.5:
    return "Excellent"
  elif average >= 7:
    return "Good"
  elif average >= 5:
    return "Average"
  else:
    return "Fail"

df["Level"] = df["Average"].apply(classify_student)
# print(df)

# Require 6: Passed
df["Passed"] = df["Average"].apply(lambda x: x >= 5)
print(df)

# Require 7: Average by City
print(df.groupby("City")["Average"].mean())

# Require 8: Subject Average by City
print(df.groupby("City")[["Math", "English", "AI"]].mean())

# Require 9: City Statistics
print(df.groupby("City")["Average"].agg(["mean", "min", "max"]))

# Require 10: Level Count
print(df["Level"].value_counts())

# Require 11: Best City
city_average = df.groupby("City")["Average"].mean()

best_city = city_average.idxmax()
print(best_city)

# Require 12: Best Student per City
max_average = df.groupby("City")["Average"].max()

for city, average in max_average.items():
    best_student = df[(df["City"] == city) & (df["Average"] == average)]

    print(best_student[["City", "Name", "Average"]])

# Require 13: Sort Students
print(df.sort_values(by="Average", ascending=False)[["Name", "City", "Average", "Level"]])

# Require 14: Summary DataFrame
city_summary = df.groupby("City").agg(
    AverageScore = ("Average", "mean"),
    StudentCount = ("StudentID", "count")
).reset_index()

# Require 15: Save Result
city_summary.to_csv("student_performance_summary.csv", index=False)
````
