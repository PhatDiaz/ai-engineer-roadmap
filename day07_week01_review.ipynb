# **Task 2 — Python Coding Review**

**Ex 1: List & Loop**

```python
numbers = [3, 8, 2, 10, 5, 7]

result = [number for number in numbers if number >= 5]

print(result)
```

**Ex 2: Function**

```python
def calculate_average(numbers):
  total = 0
  for number in numbers:
    total += number
  return total / len(numbers)

numbers = [3, 8, 2, 10, 5, 7]
average = calculate_average(numbers)
print(average)
```

**Ex 3: Dictionary**

```python
students = [
    {"name": "An", "score": 8.5},
    {"name": "Binh", "score": 7.0},
    {"name": "Cuong", "score": 9.0},
    {"name": "Dung", "score": 6.0}
]

highest_student = max(students, key = lambda student: student["score"])

print(f"{highest_student['name']} - {highest_student['score']}")
```

**Ex 4: Sorting**

```python
students = [
    {"name": "An", "score": 8.5},
    {"name": "Binh", "score": 7.0},
    {"name": "Cuong", "score": 9.0},
    {"name": "Dung", "score": 6.0}
]

sorted_students = sorted(
    students,
    key = lambda student: student["score"],
    reverse = True
)

for student in sorted_students:
  print(f"{student["name"]} - {student["score"]}")
```

**Ex 5: Function + Classification**

```python
students = [
    {"name": "An", "score": 8.5},
    {"name": "Binh", "score": 7.0},
    {"name": "Cuong", "score": 9.0},
    {"name": "Dung", "score": 6.0}
]

def classify_score(score):
  if score >= 8.5:
    return "Excellent"
  elif score >= 7:
    return "Good"
  elif score >= 5:
    return "Average"
  else:
    return "Fail"

for student in students:
  print(f"{student["name"]} - {classify_score(student["score"])}")
```

**Ex 6: OOP**

```python
class Student:
  def __init__(self, name, score):
    self.name = name
    self.score = score

  def is_passed(self):
    return self.score >= 5

  def get_grade(self):
    if self.score >= 8.5:
      return "Excellent"
    elif self.score >= 7:
      return "Good"
    elif self.score >= 5:
      return "Average"
    else:
      return "Fail"

  def display_info(self):
    print(f"{self.name} - {self.score} - {self.get_grade()}")
```

**Ex 7: Exception Handling**

```python
def safe_divide(a, b):
  try:
    return a / b

  except ZeroDivisionError:
    print("Cannot divide by zero")
```

# **Task 3 — NumPy Coding Review**

```python
import numpy as np

scores = np.array([
    [8.5, 7.0, 9.0],
    [6.0, 8.0, 7.5],
    [9.0, 9.5, 8.5],
    [5.5, 6.0, 7.0]
])

# Ex 8: Array Properties
print(np.shape(scores)) # in ra kích thước của mảng
print(np.ndim(scores))  # in ra số chiều của mảng
print(np.size(scores))  # in ra số lượng phần tử của mảng
print(scores.dtype)     # in ra kiểu dữ liệu của mảng

# Ex 9: Indexing
print(scores[2][1])

# Ex 10: Slicing
print(scores[:, 2])

# Ex 11: Statistics
print(np.mean(scores))
print(np.min(scores))
print(np.max(scores))
print(np.std(scores))

# Ex 12: Axis
print(np.mean(scores, axis = 0))
print(np.mean(scores, axis = 1))

# Ex 13: Best Student
best_student_index = np.argmax(np.mean(scores, axis = 1))
print(f"Student {best_student_index + 1}")

# Ex 14: Boolean Indexing
scores_1d = np.array([
    8.5,
    4.0,
    7.5,
    3.0,
    9.0
])

print(scores_1d[scores_1d >= 5])

# Ex 15: Weighted Score
weights = np.array([
    0.3,
    0.2,
    0.5
])

weighted_scores = scores @ weights
print(weighted_scores)
weighted_scores_index = np.argmax(weighted_scores)
print(f"Student {weighted_scores_index + 1}")
```

# **Task 4 — Final Week 1 Challenge**

```python
import numpy as np

students = [
    "An",
    "Binh",
    "Cuong",
    "Dung",
    "Hoa"
]

subjects = [
    "Math",
    "English",
    "AI"
]

scores = np.array([
    [8.5, 7.0, 9.0],
    [6.0, 8.0, 7.5],
    [9.0, 9.5, 8.5],
    [5.5, 6.0, 7.0],
    [7.5, 8.5, 8.0]
])

# Require 1: Student Average
print("Student Averages")
print("---------------- \n")

average_score_students = np.mean(scores, axis = 1)

for student, average in zip(students, average_score_students):
  print(f"{student}: {average:.2f}")
print("\n")

# Require 2: Subject Average
print("Subject Averages")
print("---------------- \n")

average_score_subjects = np.mean(scores, axis = 0)

for subject, average in zip(subjects, average_score_subjects):
  print(f"{subject}: {average:.2f}")
print("\n")

# Require 3: Best Student
print("Best Student")
print("---------------- \n")

best_student_index = np.argmax(average_score_students)
print(f"Best student: {students[best_student_index]}")
print(f"Average: {average_score_students[best_student_index]:.2f}")

# Require 4: Best Subject
print("Best Subject")
print("---------------- \n")

best_subject_index = np.argmax(average_score_subjects)
print(f"Best subject: {subjects[best_subject_index]}")
print(f"Average: {average_score_subjects[best_subject_index]:.2f}")
print("\n")

# Require 5: Classification
print("Classification")
print("---------------- \n")

def classify_score(score):
  if score >= 8.5:
    return "Excellent"
  elif score >= 7:
    return "Good"
  elif score >= 5:
    return "Average"
  else:
    return "Fail"

for student, score in zip(students, average_score_students):
  print(f"{student} - {classify_score(score)}")
print("\n")

# Require 6: Passed Students
print("Passed Students")
print("---------------- \n")

print("Passed Students:")
passed_students = [student for student, average in zip(students, average_score_students) if average >= 5]

for student in passed_students:
  print(student)
print("\n")

# Require 7: Failed Students
print("Failed Students")
print("---------------- \n")

failed_students = [student for student, average in zip(students, average_score_students) if average < 5]

if not failed_students:
  print("No failed students")
else:
  print("Failed Students:")
  for student in failed_students:
    print(student)
print("\n")

# Require 8: Statistics Function
print("Statistics Function")
print("---------------- \n")

def get_statistics(scores):
  return {
    "average": float(np.mean(scores)),
    "highest": float(np.max(scores)),
    "lowest": float(np.min(scores)),
    "std": float(np.std(scores))
  }

print(get_statistics(scores))
print("\n")

# Require 9: Weighted Score
print("Weighted Score")
print("---------------- \n")

weights = np.array([
    0.3,
    0.2,
    0.5
])

weighted_scores = scores @ weights
for student, score in zip(students, weighted_scores):
  print(f"{student}: {score:.2f}")
print("\n")

# Require 10: Best Weighted Student
print("Best Weighted Student")
print("---------------- \n")

print(f"Best weighted Student: {students[np.argmax(weighted_scores)]}")
print(f"Weighted Score: {np.max(weighted_scores):.2f}")
print("\n")

# Require 11: Search Student
print("Search Student")
print("---------------- \n")

def find_student(students, name):
  for index, student in enumerate(students):
    if name == student:
      return index
  return -1

print(find_student(students, "Huy"))
print("\n")

# Require 12: Student Report
print("Student Report")
print("---------------- \n")

def student_report(students, scores, student_name):
  print("===== STUDENT REPORT =====")

  if not student_name in students:
    print("Student not found")
  else:
    student_index = find_student(students, student_name)
    student_scores = scores[student_index]
    student_average = np.mean(student_scores)
    student_classification = classify_score(student_average)

    print(f"Name: {student_name}\n")

    for subject, score in zip(subjects, student_scores):
      print(f"{subject}: {score:.2f}")

    print("\n")
    print(f"Average: {student_average:.2f}")
    print(f"Classification: {student_classification}")

student_report(students, scores, "An")
print("\n")
```
