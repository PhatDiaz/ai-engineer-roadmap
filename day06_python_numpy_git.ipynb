# **Task 2 — Kết hợp Python + NumPy**

````python
import numpy as np

scores = np.array([
    [8.5, 7.0, 9.0],
    [6.0, 8.0, 7.5],
    [9.0, 9.5, 8.5],
    [5.5, 6.0, 7.0]
])

student_averages = np.mean(scores, axis=1)

for index, average in enumerate(student_averages):
    if average >= 8:
        print(f"Student {index + 1}: Good")
    else:
        print(f"Student {index + 1}: Needs improvement")
````

Ở đây chúng ta kết hợp:
```
np.array()         np.mean()
→ NumPy

for          if         print         f-string
→ Python
```

**enumerate() là gì?**

````python
# Thay vì:
for average in student_averages:
  print(average)

# Ta có thể cần cả: index + value
for index, average in enumerate(student_averages):
  print(index, average)
````

# **Task 3 — Coding Practice**

````python
# Ex 1

scores = np.array([8.5, 7.0, 9.0, 6.0, 8.0])

average_score = np.mean(scores)
print(average_score)

lowest_score = np.min(scores)
print(lowest_score)

highest_score = np.max(scores)
print(highest_score)

standard_deviation = np.std(scores)
print(standard_deviation)

print(np.argmax(scores))
print(np.argmin(scores))

def analyze_scores(scores):
  return {
      "average": float(np.mean(scores)),
      "lowest": float(np.min(scores)),
      "highest": float(np.max(scores)),
      "standard_deviation": float(np.std(scores))
  }

print(analyze_scores(scores))
````

````python
# Ex 4

scores = np.array([
    [8.5, 7.0, 9.0],
    [6.0, 8.0, 7.5],
    [9.0, 9.5, 8.5],
    [5.5, 6.0, 7.0]
])

average_scores = np.mean(scores, axis = 1)
print(average_scores)

highest_score = np.max(average_scores)
print(highest_score)

print(np.argmax(average_scores) + 1)

def get_best_student(scores):
  average_scores = np.mean(
    scores,
    axis=1
  )

  best_index = np.argmax(
    average_scores
  )

  return (
    best_index + 1,
    average_scores[best_index]
  )

student_number, average = get_best_student(scores)

print(f"Student {student_number}")
print(f"{average:.2f}")

weights = np.array([0.3, 0.2, 0.5])

weighted_scores = scores @ weights
print(weighted_scores)

print(f"Student {np.argmax(weighted_scores) + 1} has the highest weighted score")
````

# **Task 4 — Mini Project**

````python
import numpy as np

students = [
    "An",
    "Binh",
    "Cuong",
    "Dung"
]

scores = np.array([
    [8.5, 7.0, 9.0],
    [6.0, 8.0, 7.5],
    [9.0, 9.5, 8.5],
    [5.5, 6.0, 7.0]
])

subjects = [
    "Math",
    "English",
    "AI"
]

print("===== STUDENT ANALYTICS ===== \n")

# Require 1: Student averages
print("Student Averages")
print("---------------- \n")

average_scores_students = np.mean(scores, axis=1)
for student, average in zip(students, average_scores_students):
  print(f"{student} - {average:.2f}")

# Require 2: Subject averages
print("\nSubject averages")
print("---------------- \n")

average_scores_subjects = np.mean(scores, axis=0)
for subject, average in zip(subjects, average_scores_subjects):
  print(f"{subject}: {average:.2f}")

# Require 3: Best student
print("\nBest student")
print("---------------- \n")

best_student_average = np.max(average_scores_students)
best_student_index = np.argmax(average_scores_students)

print(f"Best student: {students[best_student_index]}")
print(f"Average: {best_student_average}")

# Require 4: Best Subject
print("\nBest subject")
print("---------------- \n")

best_subject_average = np.max(average_scores_subjects)
best_subject_index = np.argmax(average_scores_subjects)

print(f"Best subject: {subjects[best_subject_index]}")
print(f"Average: {best_subject_average}")

# Require 5: Classification
print("\nClassification")
print("---------------- \n")

def classify_student(average):
  if average >= 8.5:
    return "Excellent"
  elif average >= 7:
    return "Good"
  elif average >= 5:
    return "Average"
  else:
    return "Fail"

for index, student in enumerate(students):
  print(f"{student} - {classify_student(average_scores_students[index])}")

# Require 6: Passed Students
print("\nPassed Students")
print("---------------- \n")

print("Passed Students:")
for index, student in enumerate(students):
  if average_scores_students[index] >= 5:
    print(student)

# Require 7: Statistics
print("\nStatistics")
print("---------------- \n")

def get_statistics(scores):
  return {
      "overall_average": float(np.mean(scores)),
      "highest_score": float(np.max(scores)),
      "lowest_score": float(np.min(scores)),
      "standard_deviation": float(np.std(scores))
  }

print(get_statistics(scores))

# Require 8: Weighted Score
print("\nWeighted Score")
print("---------------- \n")

weights = np.array([0.3, 0.2, 0.5])
weighted_scores = scores @ weights

for index, student in enumerate(students):
  print(f"{student} → {weighted_scores[index]:.2f}")

# Require 9: Best Weighted Student
print("\nBest Weighted Student")
print("---------------- \n")

highest_weighted_score = np.max(weighted_scores)
highest_weighted_student = np.argmax(weighted_scores)

print(f"Best weighted student: {students[highest_weighted_student]}")
print(f"Weighted score: {highest_weighted_score}")
````

# **Task 5 — Git và GitHub**

Điều đầu tiên phải phân biệt:

`Git ≠ GitHub`

**Git**

Git là **Version Control System**.

Nó giúp lưu lịch sử thay đổi code.

Ví dụ:
```
Version 1
↓
Version 2
↓
Version 3
```
Mỗi lần lưu một phiên bản quan trọng gọi là:
`commit`

**GitHub**

GitHub là nền tảng online để lưu Git repository.

Ta có thể hiểu đơn giản:
```
Git
→ quản lý version

GitHub
→ lưu/share repository online
```

**Những khái niệm Git quan trọng nhất hôm nay**

Ta chỉ cần học:
```
repository
commit
push
pull
clone
branch
```
