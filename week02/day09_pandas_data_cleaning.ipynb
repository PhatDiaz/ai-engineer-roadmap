# **Task 1 — Missing Values**

````python
# Dataset
import pandas as pd
import numpy as np

data = {
    "Name": ["An", "Binh", "Cuong", "Dung", "Hoa"],
    "Age": [22, 25, np.nan, 23, 24],
    "Score": [8.5, np.nan, 9.0, 6.0, 8.0],
    "City": ["Hanoi", "HCM", "Hanoi", None, "Hanoi"]
}

df = pd.DataFrame(data)

print(df)
````

**NaN là gì?**

NaN:
`Not a Number`

Trong Pandas, nó thường đại diện cho:
`missing value`

Ngoài np.nan, Pandas cũng có thể coi:
`None`

là missing value.

**isna()**

`df.isna()`

Sẽ trả về bảng boolean:
`False / True`

Ví dụ:

`print(df["Score"].isna())`

có thể ra:
```
0    False
1     True
2    False
3    False
4    False
```

**Đếm missing values**

Đây là pattern rất quan trọng
```
df.isna().sum()
```
Ví dụ:
```
Name     0
Age      1
Score    1
City     1
```
Tư duy:
```
isna()
↓
True / False
↓
sum()
↓
đếm True
```
Trong Python/ Pandas:
```
True  ≈ 1
False ≈ 0
```

**notna()**

Ngược với
```
isna()
```
là
```
notna()
```
Ví dụ:
```
df["Score"].notna()
```
→ những row có Score tồn tại




**Xóa missing values — dropna()**

```
clean_df = df.dropna()
```
Nó loại những row có missing value.

Ví dụ dataset trên chỉ giữ những row không thiếu dữ liệu.


**Không thay đổi DataFrame gốc nếu không gán lại**

```
df.dropna()
```
không có nghĩa là df đã thay đổi vĩnh viễn

Muốn lưu:
```
df = df.dropna()
```
hoặc:
```
clean_df = df.dropna()
```

**dropna(subset=...)**

Nếu chỉ quan tâm một column:
```
df.dropna(
    subset=["Score"]
)
```
Nghĩa là:

`Xóa những rows mà Score bị missing`

Các column khác có missing vẫn được giữ.


**Điền missing values — fillna()**

Ví dụ:
```
df["Score"] = df["Score"].fillna(0)
```
Missing Score sẽ thành 0.

Nhưng trong ML, điền 0 không phải lúc nào cũng hợp lý.


**Điền bằng mean**

Ví dụ:
```
mean_score = df["Score"].mean()

df["Score"] = df["Score"].fillna(
    mean_score
)
```
Nếu Score:
```
8.5
NaN
9.0
6.0
8.0
```
mean được tính từ các giá trị không missing.

Sau đó NaN được thay bằng mean.

**Điền String**

Ví dụ:
```
df["City"] = df["City"].fillna("Unknown)
```
Đây thường dùng cho categorical data


**Khi nào dropna, khi nào fillna?**

Ta chỉ cần hiểu:
```
dropna()
→ loại bỏ dữ liệu bị thiếu

fillna()
→ giữ row lại nhưng thay giá trị thiếu
```

Việc chọn cái nào phụ thuộc dataset và bài toán

# **Task 2 — Duplicate Data**

````python
# Dataset

data = {
    "Name": ["An", "Binh", "Cuong", "An"],
    "Score": [8.5, 7.0, 9.0, 8.5]
}

df = pd.DataFrame(data)

# Row của An bị lặp
````

**duplicated()**

```
df.duplicated()
```
Output có thể:

```
0    False
1    False
2    False
3    True
```
Row thứ 4 trùng row đầu

**Đếm duplicate**

````python
df.duplicated().sum()
# -> 1
````

**Xóa duplicate**

````python
df.drop_duplicates()
````

**Duplicate theo một column**

Ví dụ:
```
df.duplicated(subset=["Name"])
```
Nó chỉ xét column Name

Tương tự:
```
df.drop_duplicates(subset=["Name"])
```

**keep**

Pandas mặc định:
```
df.drop_duplicate(keep="first")
```

nghĩa là:

`giữ row đầu tiên, xóa duplicate sau đó.`

Optional hôm nay:
```
keep="last"
```
-> giữ row cuối


# **Task 3 — Filtering**

````python
data = {
    "Name": ["An", "Binh", "Cuong", "Dung", "Hoa"],
    "Age": [22, 25, 21, 23, 24],
    "Score": [8.5, 7.0, 9.0, 6.0, 8.0],
    "City": ["Hanoi", "HCM", "Hanoi", "Danang", "Hanoi"]
}

df = pd.DataFrame(data)

# Filter Score >= 8
# df["Score"] >= 8

# Giống NumPy Boolean Indexing mà ta đã học
df[df["Score"] >= 8]
````

**Tư duy filtering**
```
df["Score"] >= 8
        ↓
Boolean Series
        ↓
df[Boolean Series]
        ↓
chỉ lấy rows True
```

Đây chính là Boolean Indexing phiên bản Pandas

````python
# Filter String

df[df["City"] == "Hanoi"]
# chỉ students ở Hanoi
````

````python
# Filter khác

df[df["City"] != "Hanoi"]
# chỉ students không ở Hanoi
````

````python
# Filter khoảng

df[df["Age"] > 22]
````

# **Task 4 — Multiple Conditions**

**AND**

Muốn
```
Score >= 8
AND
City == Hanoi
```
Dùng:
```
df[
    (df["Score"] >= 8)
    &
    (df["City"] == "Hanoi")
]
```




````python
# OR
df[(df["Score"] >= 9) | (df["City"] == "HCM")]

#  NOT
df[~(df["City"] == "Hanoi")]

# nhưng để dễ đọc hơn, có thể dùng:
df[df["City"] != "Hanoi"]
````

**Rất quan trọng: Pandas không dùng and / or**

Sai:
```
(df["Score"] >= 8)
and
(df["City"] == "Hanoi")
```
Đúng:
```
& → AND
| → OR
~ → NOT
```
Vì chúng ta đang làm việc với Series boolean, không phải một boolean duy nhất

````python
# Filter và chọn columns cùng lúc
high_scores = df[df["Score"] >= 8]

print(high_scores[["Name", "Score"]])
````

# **Task 6 — Coding Practice**

````python
# Dataset

import pandas as pd
import numpy as np

data = {
    "Name": ["An", "Binh", "Cuong", "Dung", "Hoa", "Huy", "An"],

    "Age": [22, 25, np.nan, 23, 24, 26, 22],

    "Score": [8.5, np.nan, 9.0, 6.0, 8.0, 7.5, 8.5],

    "City": ["Hanoi", "HCM", "Hanoi", "Danang", "Hanoi", None, "Hanoi"]
}

df = pd.DataFrame(data)
````

````python
# Ex 1: Ex 1 — Check Missing
df.isna()
````

````python
# Ex 2: Count Missing
df.isna().sum()
````

````python
# Ex 3: Missing Score
# Lấy những rows có Score missing
print(df[df["Score"].isna()])
````

````python
# Ex 4: Non-Missing Score
df[df["Score"].notna()]
````

````python
# Ex 5: Fill Score
mean_score = df["Score"].mean()

df["Score"] = df["Score"].fillna(mean_score)

print(df)
````

````python
# Ex 6: Fill City
df["City"] = df["City"].fillna("Unknown")

print(df)
````

````python
# Ex 7: Check Duplicates
print(df.duplicated())
print(df.duplicated().sum())
````

````python
# Ex 8: Remove Duplicates
df = df.drop_duplicates()
print(df.duplicated().sum())
````

````python
# Ex 9: Filter Score
df[df["Score"] >= 8]
````

````python
# Ex 10: Filter City
df[df["City"] == "Hanoi"]
````

````python
# Ex 11: Multiple Conditions
df[(df["Score"] >= 8) & (df["City"] == "Hanoi")]
````

````python
# Ex 12: OR Condition
df[(df["Score"] >= 9) | (df["Age"] >= 25)]
````

````python
# Ex 13: Sort Score Ascending
df.sort_values(by = "Score")
````

````python
#  Ex 14: Sort Score Descending
df.sort_values(by = "Score", ascending=False)
````

````python
#  Ex 15: Filter + Sort
high_scores = df[df["Score"] >= 7]

high_scores = high_scores.sort_values(by = "Score", ascending=False)

print(high_scores[["Name", "Score", "City"]])
````

# **Task 7 — Mini Challenge**
# **Student Data Cleaning Pipeline**

````python
# Dataset

import pandas as pd
import numpy as np

data = {
    "StudentID": [101, 102, 103, 104, 105, 106, 102],

    "Name": ["An", "Binh", "Cuong", "Dung", "Hoa", "Huy", "Binh"],

    "Age": [22, 25, 21, np.nan, 24, 26, 25],

    "Math": [8.5, 6.0, 9.0, 5.5, np.nan, 8.0, 6.0],

    "English": [7.0, 8.0, 9.5, 6.0, 8.5, np.nan, 8.0],

    "AI": [9.0, 7.5, 8.5, 7.0, 8.0, 9.5, 7.5],

    "City": ["Hanoi", "HCM", "Hanoi", "Danang", None, "HCM", "HCM"]
}

df = pd.DataFrame(data)
````

````python
# Require 1: Overview
print(df.head())

print(df.shape)

df.info()
````

````python
# Require 2: Missing Values
print(df.isna().sum())
````

````python
# Require 3: Fill Numeric Missing
df["Age"] = df["Age"].fillna(df["Age"].mean())
df["Math"] = df["Math"].fillna(df["Math"].mean())
df["English"] = df["English"].fillna(df["English"].mean())

print(df)
````

````python
# Require 4: Fill City
df["City"] = df["City"].fillna("Unknown")

print(df)
````

````python
# Require 5: Verify Missing
print(df.isna().sum())
````

````python
# Require 6: Detect Duplicate
print(df.duplicated().sum())
````

````python
# Require 7: Remove Duplicate
df = df.drop_duplicates()

print(df)
````

````python
# Require 8: High Performers

high_performers = df[df["AI"] >= 8]
print(high_performers[["Name", "AI"]])
````

````python
# Require 9: Hanoi Students
print(df[df["City"] == "Hanoi"])
````

````python
# Require 10: Strong Students
strong_students = df[(df["Math"] >= 8) & (df["AI"] >= 8)]

print(strong_students[["Name", "Math", "AI"]])
````

````python
# Require 11: Multiple Criteria
result = df[(df["English"] >= 8) | (df["AI"] >= 9)]

print(result)
````

````python
# Require 12: Sort AI
print(df.sort_values(by = "AI", ascending=False))
````

````python
# Require 13: Sort Math
print(df.sort_values(by = "Math"))
````

````python
# Require 14: Clean Final Dataset
final_df = df[["StudentID", "Name", "Math", "English", "AI", "City"]]

final_df = final_df.sort_values(by = "AI", ascending=False)
print(final_df)
````

````python
# Require 15: Save Clean Data
final_df.to_csv("clean_students.csv", index=False)

print(pd.read_csv("clean_students.csv"))
````
