# **Task 1 — Matplotlib là gì?**

````python
# Thư viện phổ biến nhất để vẽ biểu đồ trong Python

import matplotlib.pyplot as plt
import pandas as pd
````

**Flow cơ bản khi vẽ biểu đồ**

Thông thường:
```
plt.plot(...)
plt.title(...)
plt.xlabel(...)
plt.ylabel(...)
plt.show()
```
Tư duy:
```
Data
↓
Chọn loại chart
↓
Thêm title / label
↓
Hiển thị
```




**plt.show()**

````python
x = [1, 2, 3, 4]
y = [10, 20, 15, 30]

plt.plot(x, y)

plt.show()
````

# **Task 2 — Line Plot**

Dùng khi dữ liệu có:
```
thứ tự
xu hướng
thời gian
progression
```



````python
# Ví dụ
days = [1, 2, 3, 4, 5]

scores = [6.0, 6.5, 7.0, 8.0, 8.5]

plt.plot(days, scores)

plt.show()
````

````python
# Thêm title và axis labels

plt.plot(days, scores)

plt.title("Score Progress")

plt.xlabel("Day")
plt.ylabel("Score")

plt.show()
````

***NHỚ:***
```
title()
→ tên biểu đồ

xlabel()
→ tên trục X

ylabel()
→ tên trục Y
```



**Marker**

````python
# Có thể điểm thêm trên line:
plt.plot(days, scores, marker="o")

plt.show()

# o o o o giúp nhìn từng data point dễ hơn
````

**Nhiều line trong cùng chart**

````python
# Ví dụ:
days = [1, 2, 3, 4, 5]

math = [6, 7, 7.5, 8, 8.5]

ai = [7, 7.5, 8, 8.5, 9]

plt.plot(days, math, marker="o", label="Math")

plt.plot(days, ai, marker="o", label="AI")

plt.legend()

plt.show()
````

`label=` đặt tên cho line

```
plt.legend()
```
hiển thị chú thích


# **Task 3 — Scatter Plot**

Scatter plot dùng khi muốn xem:

`quan hệ giữa 2 biến numerical`

````python
# Ví dụ:
study_hours = [1, 2, 3, 4, 5, 6]

scores = [5, 6, 6.5, 7.5, 8, 9]

plt.scatter(study_hours, scores)

plt.xlabel("Study Hours")
plt.ylabel("Score")

plt.title("Study Hours vs Score")

plt.show()
````

Bạn có thể nhìn:
```
Study Hours tăng
↓
Score có xu hướng tăng
```
Đây là kiểu visualization sau này rất hữu ích trong ML

**Line khác Scatter thế nào?**

Line:
```
plt.plot(x, y)
```
nối các points lại

Scatter:
```
plt.scatter(x, y)
```
chỉ hiển thị từng point

**Khi nào dùng Scatter?**

Ví dụ:
```
Age vs Salary
Height vs Weight
Study Hours vs Score
Advertising vs Revenue
Feature X vs Target Y
```
Mục tiêu:

`xem hai biến numerical có relationship hay pattern gì không`

# **Task 4 — Bar Chart**

````python
# Bar chart phù hợp khi so sánh:
# categories

# Ví dụ:
cities = ["Hanoi", "HCM", "Danang"]

student_count = [30, 25, 15]

plt.bar(cities, student_count)

plt.title("Students by City")

plt.xlabel("City")
plt.ylabel("Number of Students")

plt.show()
````

Tư duy:
```
category
→ City

value
→ student count
```

**Bar Chart rất hợp với value_counts()**

````python
# Ví dụ:
city_counts = (df["City"].value_counts())

print(city_counts)

# Sau đó:
plt.bar(city_counts.index, city_counts.values)

plt.show()

# Ở đây
# city_counts.index → tên City

# city_counts.values → số lượng

# Ví dụ:
# Hanoi → 3
# HCM → 2
# Danang → 1
````

**Bar Chart + GroupBy**

# **Task 5 — Histogram**

Histogram khác bar chart

Histogram dùng để xem:

`distribution của numerical data.`

````python
# Ví dụ:
scores = [5, 5.5, 6, 6, 6.5, 7, 7.5, 8, 8.5, 9, 9.5]

plt.hist(scores)

plt.show()
````

Nó chia Score thành các khoảng:
```
5–6
6–7
7–8
8–9
...
```
và đếm số điểm nằm trong từng khoảng

**bins**

Ví dụ:
```
plt.hist(
    scores,
    bins=5
)
```
bins=5:

`chia range dữ liệu thành 5 khoảng`

Ví dụ concept:
```
Score range:
5 → 10

5 bins:

5–6
6–7
7–8
8–9
9–10
```
Không phải lúc nào mỗi bin cũng rộng đúng 1; đây chỉ là ví dụ để hiểu.


**Bar vs Histogram**

**Bar**
```
plt.bar(...)
```
dùng:
`categorical data`

Ví dụ:
```
Hanoi
HCM
Danang
```

**Histogram**

```
plt.hist(...)
```
dùng:
```
continuous / numerical data
```
Ví dụ:
```
Age
Score
Salary
Height
```
Histogram tự chia numerical values thành bins.

***NHỚ***
```
Bar
→ category comparison

Histogram
→ numerical distribution
```

# **Task 6 — Chart Formatting cơ bản**

**title()**

```
plt.title(
    "Student Scores"
)
```

**xlabel()**

```
plt.xlabel(
    "Student"
)
```

**ylabel()**

```
plt.ylabel("Score")
```

**legend()**

Khi có nhiều series:
```
plt.plot(x, y1, label="Math")

plt.plot(x, y2, label="AI")

plt.legend()
```

**grid()**

```
plt.grid()
```
giúp dễ đọc giá trị.

Ví dụ:

```
plt.plot(days, scores, marker="o")

plt.grid()

plt.show()
```

**figure(figsize=...)**

Có thể chỉnh kích thước chart:

```
plt.figure(figsize=(8, 5))
```
Ví dụ:
```
plt.figure(figsize=(8, 5))

plt.plot(days, scores)

plt.show()
```

Ý nghĩa:
```
8 → width

5 → height
```

# **Coding Practice — 15 bài**

````python
# Dùng dataset:
import pandas as pd
import matplotlib.pyplot as plt

students = pd.DataFrame({
    "Name": ["An", "Binh", "Cuong", "Dung", "Hoa", "Huy"],

    "Age": [22, 25, 21, 23, 24, 26],

    "Math": [8.5, 6.0, 9.0, 5.5, 7.5, 8.0],

    "English": [7.0, 8.0, 9.5, 6.0, 8.5, 7.5],

    "AI": [9.0, 7.5, 8.5, 7.0, 8.0, 9.5],

    "City": ["Hanoi", "HCM", "Hanoi", "Danang", "Hanoi", "HCM"]
})
````

````python
# Ex 1: Simple Line Plot
plt.plot(students["Name"], students["Math"])

plt.show()
````

````python
# Ex 2: Line + Marker
plt.plot(students["Name"], students["Math"], marker="o")

plt.show()
````

````python
# Ex 3: Title + Labels
plt.plot(students["Name"], students["Math"])

plt.title("Math Scores")

plt.xlabel("Student")
plt.ylabel("Math Score")

plt.show()
````

````python
# Ex 4: Two Lines

plt.plot(students["Name"], students["Math"], marker="o", label="Math")

plt.plot(students["Name"], students["AI"], marker="o", label="AI")

plt.legend()

# Ex 5: Grid
plt.grid()

plt.show()
````

````python
# Ex 6: Scatter Math vs AI
plt.scatter(students["Math"], students["AI"])

plt.title("Math vs AI")

plt.xlabel("Math")
plt.ylabel("AI")

plt.show()
````

````python
# Ex 7: Scatter Age vs AI

plt.scatter(students["Age"], students["AI"])

plt.xlabel("Age")
plt.ylabel("AI")

plt.show()
````

````python
# Ex 8: Bar Student AI Score

plt.bar(students["Name"], students["AI"])

plt.xlabel("Name")
plt.ylabel("AI")

plt.show()
````

````python
# Ex 9: Bar City Count

city_counts = students["City"].value_counts()

plt.bar(city_counts.index, city_counts.values)

plt.show()
````

````python
# Ex 10: Bar Average AI by City
city_ai = students.groupby("City")["AI"].mean()

plt.bar(city_ai.index, city_ai.values)

plt.show()
````

````python
# Ex 11: Histogram Math

plt.hist(students["Math"])

plt.show()
````

````python
# Ex 12: Histogram AI

plt.hist(students["AI"], bins=4)

plt.show()
````

````python
# Ex 13: Average Column

students["Average"] = (students["Math"] + students["English"] + students["AI"]) / 3

plt.bar(students["Name"], students["Average"])

plt.show()
````

````python
# Ex 14: Sort + Bar

sorted_students = students.sort_values(by = "Average", ascending=False)

plt.bar(sorted_students["Name"], sorted_students["Average"])

plt.show()
````

````python
# Ex 15: Combined Visualization

plt.plot(students["Name"], students["Math"], marker="o", label="Math")

plt.plot(students["Name"], students["English"], marker="o", label="English")

plt.plot(students["Name"], students["AI"], marker="o", label="AI")

plt.title("Students' Scores")

plt.xlabel("Student")
plt.ylabel("Score")

plt.legend()

plt.grid()

plt.show()
````

# **Mini Challenge — Student Performance Visualization**

````python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "StudentID": [101, 102, 103, 104, 105, 106, 107, 108],

    "Name": ["An", "Binh", "Cuong", "Dung", "Hoa", "Huy", "Lan", "Minh"],

    "Age": [22, 25, 21, 23, 24, 26, 22, 25],

    "Math": [8.5, 6.0, 9.0, 5.5, 7.5, 8.0, 6.5, 9.5],

    "English": [7.0, 8.0, 9.5, 6.0, 8.5, 7.5, 7.0, 9.0],

    "AI": [9.0, 7.5, 8.5, 7.0, 8.0, 9.5, 7.5, 9.0],

    "City": ["Hanoi", "HCM", "Hanoi", "Danang", "Hanoi", "HCM", "Danang", "Hanoi"]
})

# Require 1: Overview
print(df.head())

print(df.shape)

df.info()
````

````python
# Require 2: Overview
df["Average"] = (df['Math'] + df['English'] + df['AI']) / 3

# Require 3: Math Line Chart
plt.plot(df["Name"], df["Math"], marker = "o")

plt.title("Math Scores")

plt.xlabel("Student")
plt.ylabel("Math Score")

plt.show()
````

````python
# Require 4: Subject Comparison

plt.plot(df["Name"], df["Math"], label = "Math")

plt.plot(df["Name"], df["English"], label = "English")

plt.plot(df["Name"], df["AI"], label = "AI")

plt.title("Subject Comparison")

plt.xlabel("Student")
plt.ylabel("Score")

plt.legend()

plt.grid()

plt.show()
````

````python
# Require 5: Math vs AI Scatter
plt.scatter(df["Math"], df["AI"])

plt.xlabel("Math")
plt.ylabel("AI")

plt.show()
````

````python
# Require 6: Age vs Average Scatter
plt.scatter(df["Age"], df["Average"])

plt.xlabel("Age")
plt.ylabel("Average")

plt.show()
````

````python
# Require 7: Average by Student Bar

plt.bar(df["Name"], df["Average"])

plt.show()
````

````python
# Require 8: Sort Average Bar

sorted_df = df.sort_values(by = "Average", ascending=False)

plt.bar(sorted_df["Name"], sorted_df["Average"])

plt.show()
````

````python
# Require 9: Student Count by City

city_counts = df["City"].value_counts()

plt.bar(city_counts.index, city_counts.values)

plt.show()
````

````python
# Require 10: Average by City
city_average = df.groupby("City")["Average"].mean()

plt.bar(city_average.index, city_average.values)

plt.show()
````

````python
# Require 11: Math Distribution

plt.hist(df["Math"])

plt.show()
````

````python
# Require 12: AI Distribution

plt.hist(df["AI"], bins=4)

plt.show()
````

````python
# Require 13: Average Distribution

plt.hist(df["Average"])

plt.title("Average Distribution")

plt.xlabel("Average")
plt.ylabel("Frequency")

plt.show()
````

````python
# Require 14: Best Student Chart
max_average = df["Average"].max()

best_student = df[df["Average"] == max_average]

print(best_student[["Name", "Average"]])

plt.bar(best_student["Name"], best_student["Average"])

plt.show()
````
